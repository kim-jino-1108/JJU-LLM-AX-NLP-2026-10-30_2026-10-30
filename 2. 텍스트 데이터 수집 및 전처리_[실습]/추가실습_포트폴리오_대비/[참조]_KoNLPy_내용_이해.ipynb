## 0. 설치·불러오기
KoNLPy는 내부에서 **Java**를 사용합니다. 아래 셀이 Java 확인과 KoNLPy 설치를 함께 처리합니다.

````python
# Java 확인 도구 불러오기
import shutil

# Java가 없으면 설치 (KoNLPy는 Java가 필요함, Colab·리눅스용)
if shutil.which("java") is None:
    !apt-get -qq install -y default-jdk-headless > /dev/null

# KoNLPy 설치
!pip install -q konlpy
````

````python
# Okt 분석기 불러오기
from konlpy.tag import Okt

# 분석기 만들기
okt = Okt()
````

## 1. 형태소 나누기
`morphs()`는 문장을 형태소 단위로 나눕니다.

````python
# 분석할 문장
text = "가격도 저렴하고 품질도 만족합니다"

# 형태소로 나누기
okt.morphs(text)
````

## 2. 명사만 뽑기
`nouns()`는 명사만 골라 줍니다.

````python
# 명사만 뽑기
okt.nouns(text)
````

## 3. 품사 붙이기
`pos()`는 (형태소, 품사) 쌍으로 결과를 줍니다.

````python
# 형태소와 품사 함께 보기
okt.pos(text)
````

## 4. 품사표 보기
Okt는 품사 이름을 영어 단어로 표시합니다. `tagset`으로 뜻을 확인할 수 있습니다.

````python
# Okt 품사표 출력
okt.tagset
````

## 5. 기본형으로 바꾸기
`stem=True`를 넣으면 `저렴하고` → `저렴하다`처럼 **기본형**으로 바뀝니다. 같은 단어를 한 가지 모양으로 셀 수 있습니다.

````python
# 기본형으로 바꾸어 품사 붙이기
okt.pos(text, stem=True)
````

## 6. 구어체 정리
`norm=True`를 넣으면 `빨라욬`, `넘` 같은 구어체를 정리합니다.

````python
# 구어체가 섞인 리뷰
talk = "오늘 받은 로션 넘 좋아욬ㅋㅋㅋ"

# 정리 전
print(okt.pos(talk))

# 정리 + 기본형
print(okt.pos(talk, norm=True, stem=True))
````

## 7. 다른 분석기와 비교
KoNLPy에는 Okt 외에도 Kkma(꼬꼬마), Komoran(코모란), Hannanum(한나눔)이 있습니다. 사용법은 같고 **결과는 다릅니다.**

````python
# 다른 분석기 불러오기
from konlpy.tag import Kkma, Komoran

# 띄어쓰기 없는 문장
text2 = "아버지가방에들어가신다"

# 분석기별 결과 비교
print("Okt    :", okt.pos(text2))
print("Kkma   :", Kkma().pos(text2))
print("Komoran:", Komoran().pos(text2))
````

분석기마다 품사 이름도 다릅니다. Okt는 `Noun`, Kkma·Komoran은 `NNG`처럼 표시합니다.

## 8. 여러 문장 명사 세기
리뷰 여러 개에서 명사를 모아 많이 나온 순서로 셉니다.

````python
# 리뷰 5개
reviews = [
    "배송이 빠르고 포장이 꼼꼼해요",
    "가격 대비 품질이 좋아요",
    "배송은 느렸지만 품질은 만족합니다",
    "포장이 찢어져서 왔어요",
    "가격이 저렴하고 배송도 빨라요",
]

# 단어 세기 도구 불러오기
from collections import Counter

# 모든 리뷰의 명사 모으기
nouns = [n for r in reviews for n in okt.nouns(r)]

# 많이 나온 명사 3개
Counter(nouns).most_common(3)
````
